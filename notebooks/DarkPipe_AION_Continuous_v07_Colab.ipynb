```python
import json
from pathlib import Path

checked = Path('evidence/aion_continuous_environment_2026-08-25')
checked_report = json.loads((checked / 'report.json').read_text())
assert checked_report['decision'] == 'NO_HOLDOUT_CANDIDATE'
assert len(checked_report['holdout_confirmation']) == 8
assert not any(x['detected'] for x in checked_report['holdout_confirmation'])
print('Recibo checked: 0/8; NO_HOLDOUT_CANDIDATE')
```

```python
import shutil
from darkpipe.aion import EVIDENCE_DIRECTORY
from darkpipe.aion_continuous import confirm_continuous_candidates

evidence = Path('evidence') / EVIDENCE_DIRECTORY
target = Path('/content/darkpipe_v07_reproduction')
target.mkdir(parents=True, exist_ok=True)
shutil.copy2(checked / 'discovery_candidates.json', target)
candidate_commit = '0fa86e0906e613207ba4e69120bcc5fea5bf7949'
report = confirm_continuous_candidates(evidence, target, candidate_commit)
assert report['decision'] == checked_report['decision']
assert report['calibration'] == checked_report['calibration']
assert report['holdout_confirmation'] == checked_report['holdout_confirmation']
assert report['environment'] == checked_report['environment']
print('Reproducción de endpoints: OK')
```
