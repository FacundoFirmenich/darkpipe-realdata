It does not download raw survey images/catalogues and does not claim an exact object-level deprojection.

```python
from pathlib import Path
import subprocess, sys

REVISION = 'v012-covariance-operator-shadow'
repo = Path('/content/darkpipe-realdata-v012')
if repo.exists():
    raise FileExistsError('Repository directory already exists; restart the runtime for a clean execution.')
subprocess.run(['git', 'clone', '--depth', '1', '--branch', REVISION, 'https://github.com/FacundoFirmenich/darkpipe-realdata', str(repo)], check=True)
subprocess.run([sys.executable, '-m', 'pip', 'install', '-q', str(repo)], check=True)
```

```python
import darkpipe
assert darkpipe.__version__ == '0.12.0', darkpipe.__version__
license_text = (repo / 'LICENSE').read_text(encoding='utf-8')
assert 'GNU GENERAL PUBLIC LICENSE' in license_text
assert 'either version 3 of the License, or' in license_text
subprocess.run([sys.executable, '-m', 'pytest', str(repo / 'tests/test_covariance_operator_shadow.py'), '-q'], cwd=repo, check=True)
print('Version, GPL-3.0-or-later, source-hash, covariance and operator gates: PASS')
```

```python
output = Path('/content/darkpipe_v012_result')
if output.exists():
    raise FileExistsError('Output already exists; restart the runtime to avoid relabeling a rerun.')
subprocess.run([sys.executable, str(repo / 'run_darkpipe_covariance_operator_v012.py'), '--output-dir', str(output)], cwd=repo, check=True)
```

```python
import json
from IPython.display import Image, Markdown, display

summary = json.loads((output / 'summary.json').read_text(encoding='utf-8'))
display(Markdown((output / 'SUBSTANTIVE_CLOSURE_ES.md').read_text(encoding='utf-8')))
display(Image(filename=str(output / 'covariance_operator_shadow.png')))
summary
```
