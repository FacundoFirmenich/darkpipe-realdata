# DarkPipe 0.11 — multi-channel observable shadows
 
 This Colab installs the public repository, verifies version and GPL-3.0-or-later, runs the frozen 15-bin weak-lensing operator against the checked v0.10 SPARC surface, displays the substantive result, and downloads only a compact ZIP. It does not download raw KiDS survey data and does not perform object-level fusion.


```python
from pathlib import Path
 import subprocess, sys
 
 repo = Path("/content/darkpipe-realdata-v011")
 if repo.exists():
     raise FileExistsError("Repository directory already exists; restart the runtime for a clean execution.")
 subprocess.run(["git", "clone", "--depth", "1", "https://github.com/FacundoFirmenich/darkpipe-realdata", str(repo)], check=True)
 subprocess.run([sys.executable, "-m", "pip", "install", "-q", str(repo)], check=True)
```

```python
import darkpipe
 assert darkpipe.__version__ == "0.11.0", darkpipe.__version__
 license_text = (repo / "LICENSE").read_text(encoding="utf-8")
 assert "GNU GENERAL PUBLIC LICENSE" in license_text
 assert "either version 3 of the License, or" in license_text
 print("Version and GPL-3.0-or-later gate: PASS")
```

```python
output = Path("/content/darkpipe_v011_result")
 if output.exists():
     raise FileExistsError("Output already exists; restart the runtime to avoid relabeling a rerun.")
 subprocess.run([
     sys.executable,
     str(repo / "run_darkpipe_multishadow_v011.py"),
     "--output", str(output),
 ], cwd=repo, check=True)
```

```python
import json
 from IPython.display import Image, Markdown, display
 
 summary = json.loads((output / "multishadow_summary.json").read_text(encoding="utf-8"))
 display(Markdown((output / "SUBSTANTIVE_CLOSURE_ES.md").read_text(encoding="utf-8")))
 display(Image(filename=str(output / "multishadow_atlas.png")))
 summary
```
